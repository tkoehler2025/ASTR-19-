```python
import numpy as np
```

```python
from tabulate import tabulate
```

## X Values

```python
x = np.linspace(0,2*np.pi,1000)
```

## Y value takes the sin of X

```python
y = np.sin(x)
```

## Z value takes the cos of X

```python
z = np.cos(x)
```

# Data Table 

```python
def main():
    table_data = [(a, b, c) for a, b, c in zip(x, y, z)]
    headers = ["x", "sin(x)", "cos(x)"]
    table = tabulate(table_data, tablefmt="grid", headers=headers, floatfmt=".4f")
    print(table)
if __name__=="__main__":
    main()
```

```python
def main():
    table_data = [(a, b, c) for a, b, c in zip(x, y, z)]
    headers = ["x", "sin(x)", "cos(x)"]
    table = tabulate(table_data[:10], tablefmt="grid", headers=headers, floatfmt=".4f")
    print(table)
if __name__=="__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from numpy import random
```

```python
x = random.choice([10,20,30,40,50,5,15,25,35,45,2,4,6,8,12,14,16,17,18], size=(1000))
```

```python
plt.hist(x, bins=100)
plt.xlabel("difference of final scores of basket ball games in one season")
plt.ylabel("how many games had that point difference")
plt.savefig("Basket-diff_hist.pdf")
```

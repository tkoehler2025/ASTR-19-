```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(1,0,100)
```

```python
y = np.sin(7*x)
z = np.cos(7*x)
```

```python
fig, axes = plt.subplots(1,2)
axes[0].plot(x,y, label='sin(x)')
axes[1].plot(x,z, label='cos(x)')
axes[0].legend()
axes[1].legend()
plt.savefig("sin and cos.pdf")
```
